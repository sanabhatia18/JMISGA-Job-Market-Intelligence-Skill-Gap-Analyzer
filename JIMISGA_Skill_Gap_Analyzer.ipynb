```python
import pandas as pd
```

```python
df=pd.read_csv("india_job_market_2024-26.csv")
df=df.drop(columns=['Unnamed: 15','Unnamed: 16'])
df.head()
```

```python
career=input("enter the career you want to analyze: ")
career_jobs=df[df["Job_Title"] == career]
career_jobs.shape
```

```python
df["Job_Title"].unique()
```

```python
career_skills=[]
for skill_list in career_jobs["Skills_Required"]:
    for skill in skill_list.split(","):career_skills.append(skill.strip())
career_skill_counts=pd.Series(career_skills).value_counts()
career_skill_counts
```

```python
user_input=input("Enter your skills separated by commas: ")
user_skills=[]
for skill in user_input.split(","):
    user_skills.append(skill.strip())
user_skills
```

```python
required_skills=career_skill_counts.index.tolist()
required_skills
```

```python
matched_skills=[]
for skill in required_skills:
    for user_skill in user_skills:
        if skill.lower() == user_skill.lower():
            matched_skills.append(skill)
matched_skills
```

```python
missing_skills=[]
for skill in required_skills:
    found= False

    for user_skill in user_skills:
        if skill.lower() == user_skill.lower():
            found=True

    if found == False:
        missing_skills.append(skill)
missing_skills
```

```python
match_percentage=(len(matched_skills) / len(required_skills)) *100
match_percentage=round(match_percentage,2)
print("Skill Match Percentage: ", match_percentage , "%")
```

```python
all_careers=df["Job_Title"].unique()
all_careers
```

```python
for career in all_careers:
    
    career_jobs = df[df["Job_Title"] == career]
    
    print(career, career_jobs.shape)
```

```python
for career in all_careers:
    career_jobs=df[df["Job_Title"] == career]
    career_skills=[]
    for skill_list in career_jobs["Skills_Required"]:
        for skill in skill_list.split(","):
            career_skills.append(skill.strip())
    print(career, career_skills[:5])
```

```python
for career in all_careers:
   career_jobs=df[df["Job_Title"] == career]
   career_skills=[]
   for skill_list in career_jobs["Skills_Required"]:
      for skill in skill_list.split(","):
         career_skills.append(skill.strip())
   career_skill_counts=pd.Series(career_skills).value_counts()
   print("\n",career)
   print(career_skill_counts.head(7))
```

```python
career_core_skills={}
for career in all_careers:
    career_jobs = df[df["Job_Title"] == career]
    career_skills = []
    for skill_list in career_jobs["Skills_Required"]:
        for skill in skill_list.split(","):
            career_skills.append(skill.strip())
    career_skill_counts = pd.Series(career_skills).value_counts()
    top_skills = career_skill_counts.head(7)
    career_core_skills[career] = top_skills.index.tolist()
```

```python
career_core_skills["Data Scientist"]
```

```python
career_matches=[]
for career in all_careers:
    required=career_core_skills[career]
    print(career,
          required)
```

```python
user_skills_lower = []
for skill in user_skills:
    user_skills_lower.append(skill.lower())
print(user_skills_lower)
```

```python
for job in all_careers:
    required=career_core_skills[job]
    matched=[]
    for skill in required:
            if skill.lower() in user_skills_lower:
                matched.append(skill)
    print(job,matched)
```

```python
career_results=[]
for job in all_careers:
    required = career_core_skills[job]
    matched = []
    for skill in required:
        if skill.lower() in user_skills_lower:
            matched.append(skill)
    match_percentage=(len(matched) / len(required)) *100
    match_percentage=round(match_percentage,2)
    career_results.append([job,match_percentage])
career_results
```

```python
career_final_result=pd.DataFrame(career_results,columns=["Career","Match Percentage(%)"])
career_final_result
```

```python
career_final_result=career_final_result.sort_values(by="Match Percentage(%)",ascending=False)
career_final_result
```

```python
top_5_careers=career_final_result.head(5)
top_5_careers
```

```python
career_openings=df.groupby("Job_Title")["Openings"].sum()
career_openings
career_final_result['Total Openings']=career_final_result["Career"].map(career_openings)
fresher_jobs=df[df["Experience_Level"]=="Fresher (0-1 yr)"]
fresher_openings=fresher_jobs.groupby("Job_Title")["Openings"].sum()
career_final_result["Fresher Openings"] =career_final_result["Career"].map(fresher_openings)
```

```python
average_salary=df.groupby("Job_Title")["Salary_LPA"].mean()
career_final_result["Average Salary(LPA)"]=career_final_result["Career"].map(average_salary)
career_final_result["Average Salary(LPA)"]=career_final_result["Average Salary(LPA)"].round(2)
career_final_result.head(5)
```

```python
matched_skills_all={}
missing_skills_all={}
for job in all_careers:
    required=career_core_skills[job]
    matched=[]
    missing=[]
    for skill in required:
        if skill.lower() in user_skills_lower:
            matched.append(skill)
        else:
            missing.append(skill)
    matched_skills_all[job]=matched
    missing_skills_all[job]=missing
```

```python
print(matched_skills_all.keys())
```

```python
career_final_result["Matched Skills"]=career_final_result["Career"].map(matched_skills_all)
career_final_result["Missing Skills"]=career_final_result["Career"].map(missing_skills_all)
career_final_result.head(5)
```

```python
simulated_match={}
for job in all_careers:
    required=career_core_skills[job]
    matched=[]
    for skill in required:
        if skill.lower() in simulated_skills_lower:
            matched.append(skill)
    match_percentage=(len(matched)/len(required))*100
    simulated_match[job]=round(match_percentage,2)
```
